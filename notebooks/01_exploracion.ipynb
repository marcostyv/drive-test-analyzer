```python
'''
EJERCICIO 3: Comprueba los valores de RSRP de todos los ficheros para ver si hay algo extraño
'''

from pathlib import Path ## Herramienta para trabajar con rutas, ficheros, carpetas.

tablas = [] ## Crea una lista vacía. Las listas son mutables, de distintos elementos, ordenadas por índice...
for ruta in Path("../data/raw/5G-production-dataset").rglob("*.csv"): ## Esa función busca todos los csv dentro del directorio y subdirectorios, devuele una lista y "ruta" recorre esa lista en el bucle for.
    nuevo_csv = pd.read_csv(ruta, na_values='-')  ## Cargar el fichero.
    tablas.append(nuevo_csv) ## Simplemente añade al final.

todo = pd.concat(tablas)  ## Coge todos los csv y los junta en un dataframe uno debajo de otro... Así tenemos todos los datos de todos los csv a la vez
todo["RSRP"].describe() ## Filtramos solo la columna de RSRP y vemos los valores maximo, minimo, desviacion, media...

# '''
# RESPUESTA: Sí, hay valores de -200 dB, lo cual es imposible de medir / no especificado por el 3GPP. Si vamos al 3GPP (ver doc) vemos que efectivamente son valores imposibles. Es de -44 dBm a -140 dBm.
# '''
```
