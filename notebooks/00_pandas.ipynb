## Explicación del **.loc** y la diferencia con **.iloc**

Ambos sirven para acceder a filas y columnas del modo **metodo[FILA, COLUMNA]** pero de distinta forma

**.loc** accede por Etiqueta siempre. Ejemplos:

```python
df.loc[0] ## Me muestra la FILA de etiqueta '0'. En el dataframe recién extraido del csv, las etiquetas son los números, por eso sí funciona.
df.head(1) ## Coinciden los valores.

df.loc[0,0] ## NO funciona porque no hay ninguna columna con Etiqueta 0!
```

```python
df.loc[800,"RSRP"] ## Ahora sí
df.loc[[3,4,5], ["RSRP", "Timestamp"]] ## Puedo crearme mi propia tabla con las etiquetas de fila y columna. RECORDAR poner en una lista los valores!
```

Sobre **.iloc** es lo mismo pero por valor numérico

```python
df.iloc[0] ## Accede a la fila de posición 0!
df.iloc[0,0] ## Fila posicion 0, columna posicion 0.

## El .iloc puede ser más útil en columnas tras operaciones de cosas, donde sabes que la posción tiene algo que ver. por ejemplo al ordenar una tabla de menor a mayor por cierto parámetro... etc
```

Ejercicio 3: filtrar con máscaras
- a) ¿Cuántas muestras tienen RSRP < −100 dBm? ¿Qué porcentaje del total son?
- b) De esas, ¿cuántas tienen además SNR < 0 dB? Es decir, mala cobertura y mala calidad a la vez.
- c) Muestra esas filas, solo con Timestamp, RSRP, SNR y CQI.

Nuevo: filtrar la tabla con una máscara, df[mascara], en vez de usarla para asignar.

```python
''' Ejercicio 3a: ¿Cuántas muestras tienen RSRP < −100 dBm? ¿Qué porcentaje del total son?'''
RSRP_bajo = (df["RSRP"] < -100).sum() ## Cogemos la columna y evaluamos cuales son menor a -100. Esta lista de True y False hacemos un sum (recuerda suma 1 los Trues 0 los falses)
print("Muestras de RSRP menores a -100 dBm: " + str(RSRP_bajo))
RSRP_bajo_porcentaje = ((df["RSRP"] < -100).value_counts(normalize=True) * 100).loc[True] ## Normalizo las valores del value_counts y saco el valor en la etiqueta True (Que son los que cumplian esa condición)
RSRP_bajo_porcentaje = round(RSRP_bajo_porcentaje,2)
print("Porcentaje de muestras con RSRP menor a -100 dBm: " + str(RSRP_bajo_porcentaje) + "%")
```

```python
### HAY UNA MANERA MUCHO MEJOR DE HACER ESTO!

((df["RSRP"] < -100).mean() * 100).round(2)
### Si evalúo la condición tengo una lista de Trues y Falses. Si hago la media de esa lista ".mean()", estoy sacando el porcentaje de los que sí cumplen! porque es sumer todos los '1' (los True) y dividir entre el total!
```

**Para todas las veces en las que evalúo 1 sola condición (Una lista de True o Falses), saber que el operador ".mean()" puede ser muy útil en sacar porcentajes de forma muy sencilla !!!!**

```python
''' Ejercicio 3b: De esas, ¿cuántas tienen además SNR < 0 dB? Es decir, mala cobertura y mala calidad a la vez.'''
## Tenemos que evaluar 2 condiciones:
muestras = (df["RSRP"] < -100) & (df["SNR"] < 0) ## Simplemente cada una en su paréntesis y unidas con los signos de operaciones Bool. LOS SIGNOS!!! (NO se puede poner "and", "or", etc. eso es python nativo, no pandas)
print("Muestars con RSRP < -100 dBm y SNR < 0 dB: " + str(muestras.sum())) ## Recordemos que el sum sumará los True!
```

```python
''' Ejercicio 3c: Muestra esas filas, solo con Timestamp, RSRP, SNR y CQI.'''
mascara = (df["RSRP"] < -100) & (df["SNR"] < 0) ## Esto es la MASCARA, una condición, una lista de trues y falses evaluados por una condición.
df.loc[mascara, ["Timestamp","RSRP","SNR","CQI"]] ## Al listar las filas y columnas, podemos poner una máscara. Las que cumplan la condición serán filas que se impriman (las True), las que no no se imprimen (las False).
## Y después simplemente cogemos las columnas que queremos ver...
```

# Ejercicio 4: agrupar con groupby
- a) Calcula el RSRP medio por NetworkMode.
- b) Lo mismo, pero con RSRP y SNR a la vez.
- c) ¿Qué tecnología tiene mejor calidad media? ¿Cuadra con lo que esperarías?

Nuevo: df.groupby("columna")["otra"].mean() agrupa las filas por el valor de una columna y calcula algo para cada grupo. Para varias columnas, pasa una lista en vez de un nombre.

```python
''' Ejercicio 4a: Calcula el RSRP medio por NetworkMode.'''

df.groupby("NetworkMode").groups ## Esto sería para ver todos los grupos que se han hecho
df.groupby("NetworkMode").get_group("5G") ## Me da un grupo concreto... Pero mejor poner una mascara no? el Groupby no es tan util para esto...

df.groupby("NetworkMode") ## El groupby lo que hace es crear varios dataframe por cada columna. Por lo tanto, aquí tendríamos varios df por RAT.
df.groupby("NetworkMode")["RSRP"] ## Aquí lo que hago es que, de los distintos df que tengo, me quedo con la culumna RSRP de los mismos.
df.groupby("NetworkMode")["RSRP"].mean() ## Finalmente, calculo la media del RSRP a cada columna.
```

```python
''' Ejercicio 4b: Lo mismo, pero con RSRP y SNR a la vez.'''

df.groupby("NetworkMode")[["RSRP","SNR"]].mean() ## Podemos poner varios a la vez perfectamente, sin rayarse.
```

```python
''' Ejercicio 4c: ¿Qué tecnología tiene mejor calidad media? ¿Cuadra con lo que esperarías?'''
## Esto es aplicar inteligencia a lo que hago. Como ingeniero de redes móviles tengo que saber que es mejor utilizar la mediana. O saber que en el año que se hizo este dataset, el 5G era solo NSA (usado en condiciones óptimas),
## lo que puede sesgar medidas buenas para el 5G y no tan buenas para el LTE. O comprobar el número de muestras de cada RAT para ver si son suficientes o no (recordemos que el dataset viene de G-NetTrack y la app hace lo que
## quiera, hay que comprobar el dataset siempre!) etc.

## 1 - Evaluación inicial (el dataset puede estar mal)
df.dtypes ## Primero vemos que KPI de calidad hay en el dataset. Vemos que el RSCP o Ec/Io no aparecen (los de 3G). Bueno, seleccionaremos los otros.
df.groupby("NetworkMode")[["RSRP","RSRQ","SNR"]].median() ## Evaluamos a simple vista. Vemos que hay valores raros en HSPA y UMTS (encadenado con lo de antes). Podemos ir probando cosas:

## 2 - Comprobaciones de cosas (el dataset puede estar mal)
df.groupby("NetworkMode").size() ### ESTO nos da que UMTS solo tiene 4 muestras válidas, por lo que ni la tendremos en cuenta.
df.loc[df["NetworkMode"] == "HSPA+", "RSRQ"].value_counts() ## Para las muestras de HSPA+, vemos que los valores de RSRQ son casi siempre de "-2" y no fluctúan. Estas medidas están mal, tampoco vamos a evaluar HSPA+

## 3 - Sacamos dataset final que sí nos sirve
mascara = (df["NetworkMode"] != "UMTS") & (df["NetworkMode"] != "HSPA+") ## Podemos crear esta máscara / condición y aplicarla después.
df[mascara].groupby("NetworkMode")[["RSRP","RSRQ","SNR"]].median() ## Yo cogería las KPI Radio que más me interesan para verlo. 
## Como ves, puedo poner df[mascara] y ya me está dando la lista de Treu y Falses para, con las siguiente funciones, filtrar lo que quiero y lo que no.
```

# Ejercicio 5: crear columnas y ordenar
- a) Crea una columna nueva DL_Mbps a partir de DL_bitrate, suponiendo que DL_bitrate está en kbps. La unidad exacta la confirmarás en el ejercicio 6 de la guía.
- b) Muestra las 5 muestras con más throughput, con Timestamp, RSRP, SNR, CQI y DL_Mbps.
- c) Mira esas 5 filas: ¿cómo son el SNR y el CQI cuando el throughput es máximo? ¿Tiene sentido físicamente?

```python
''' Ejercicio 5a: Crea una columna nueva DL_Mbps a partir de DL_bitrate, suponiendo que DL_bitrate está en kbps. La unidad exacta la confirmarás en el ejercicio 6 de la guía.'''
df["DL_Mbps"] = df["DL_bitrate"] / 1000 ## Para crear una nueva columna simplemente le damos un valor "haciendo como que ya existe"
df["Operador"] = "Irlanda" ## Esto le pondría el mismo valor a todas...
df["DL_Mbps"].head()
```

```python
## También se pueden añadir filas...
nuevas = pd.DataFrame({"RSRP": [-90, -95], "SNR": [5.0, 3.0]}) ## pd.DataFrame() para crear el tipo de dato. Dentro ve un diccionario (entre corchetes), le pones la cantidad de valores que quieres.
## Ahora, concatenamos este pequeño df al df completo:
df = pd.concat([df, nuevas], ignore_index=True) ## Con esto concatenamos. Tenemos que poner lo de "ignore_index=True" para que los indices se renumeren desde 0. si no podria poner indice "1, 2" en la última fila...
df.tail() ## Como vemos se han añadido al final!
```

```python
''' Ejercicio 5b: Muestra las 5 muestras con más throughput, con Timestamp, RSRP, SNR, CQI y DL_Mbps.'''
## Primero saco el df que quiero:
df_5b = df[["Timestamp","RSRP", "SNR","CQI","DL_Mbps"]]
df_5b_ordenado_THR = df_5b.sort_values("DL_Mbps",ascending=False) ## Otra función MUY interesante, la de sort_values. Podemos invertir el orden con el parametro Ascending...
df_5b_ordenado_THR.head()
```

```python
'''Ejercicio 5c: Mira esas 5 filas: ¿cómo son el SNR y el CQI cuando el throughput es máximo? ¿Tiene sentido físicamente?'''
## Son valores muy altos, tiene todo el sentido. Cuando las condiciones del canal son óptimas, se usan modulaciones mayores, codificacione sde redundancia menores, más MIMO hacia el usuario y otras muchas técnicas. 
## Esto resulta en mayor throughput.
```

```python
import pandas as pd
notas = pd.Series([3, 5, 7, 9, 10]) ## Esto es una Series. Es como "una columna sola" de un Pandas dataframe. En este caso no tiene ni nombre de datos,
                                    ## ni nombre de filas... nada
print(notas)

## Lo que hacemos es sustituir esos valores por otros valores según un rango!
## Para ello usa la función "pd.cut(columna, bin=lista_de_límites, label=lista_de_valores_nuevos, right=True/False)". Lo último es para incluir o no
## el valor de los bins en el rango de la izquierda o en el de la derecha
pd.cut(notas, bins=[0,5,7,9,float("inf")], labels=["Insuficiente", "Suficiente", "Notable", "Sobresaliente"], right=False)
## "bins" va de menor a mayor y en el último mejor poner infinito (porque sacar un 10 sería fuera de rango al tener el right = False). Piensa
## el ejemplo y lo entenderás.
```
