Los directorios pueden ser **aboslutos** cuando la ruta empieza desde la base (C:\Users\...) o **relativos** cuando la ruta empieza desde aquí.
En caso de ser **relativos**, para comenzar a escribir la ruta tenemos:
- **"."** indica la carpeta actual. También puedes no ponerlo. "data/raw" y "./data/raw" son exactamente lo mismo.
- **".."** indica la carpeta inmediatamente superior. Se pueden concatenar varios. "../data/raw/.." o "../../5gdataset/.."

**Por cierto**, en Python y porgramación se suele poner este **"/"** pero en windows terminal pueden usarse los dos: tanto **"/"** como **"\\"**

## 1) El objeto **Path**

```python
datos = Path("../data/raw/5G-production-dataset") ## Esto crea un objeto Path, con el que se pueden hacer cosas de directorios de forma mucho más sencilla
datos
```

```python
fichero = datos / "Download" / "Driving" / "B_2019.12.14_10.16.30.csv" ## Se pueden añadir trozos de ruta con las barras "/" para crear otro Path
fichero
```

```python
## SE PUEDEN VER ATRIBUTOS DE LAS RUTAS. Cuando es un atributo directo solo se pone ".atributo" pero cuando tiene que revisarlo en disco o hacer alguna accion avanzada sí se pone ".metodo()", con el paréntesis!
print(fichero.name)      # 'B_2019.12.14_10.16.30.csv'   → nombre con extensión
print(fichero.stem)      # 'B_2019.12.14_10.16.30'       → nombre sin extensión
print(fichero.suffix)    # '.csv'                        → extensión
print(fichero.parent)    # .../Download/Driving          → la carpeta que lo contiene
fichero.exists()  # True                          → ¿existe de verdad?

## Todo esto puede servir para comprobar cosas de ficheros y rutas y crear otras cosas... ahora vemos ejercicios.
```

```python
## Otra cosa muy útil es dividirlo en todas las partes:
partes = fichero.parts ## Esto devuelve una tupla con todos los valores del directorio (carpetas, fichero final, los ".." en caso de que la ruta tenga eso...)
print(partes)
## Recordemos que en las tuplas se puede acceder por posición por corchetes:
print(partes[0]) ## Accedo a elemento 0
print(partes[1]) ## Accedo a elemento 1
print(partes[-1]) ## Accedo a elemento -1 (por el final, el último)
print(partes[-2]) ## Accedo a elemento -2 (el penúltimo)
```

```python
## BUSCAR FICHEROS
# Esto es de lo que más vamos a usar siempre con pandas y datos. busquede de muchos ficheros en directorios, etc.
datos.glob("*.csv") ## Este método lo que hace es devolver todos los archivos que acaben en csv ("*.csv" = "cualquier_cosa.csv") en el directorio actual. Pero no los devuelve en una lista, es un GENERADOR. Se tiene
## que combinar con bucles for o con list() para guardar los valores en algo u operar con ellos. El generador recorre todos 1 a 1, no repite.
list(datos.glob("*.csv")) ## Como vemos, en el directorio de 5G-production-dataset, no hay csvs. Por eso usaremos:
lista_csv = list(datos.rglob("*.csv")) ## con "r" de recursivo. busca en todos los subdirectorios!
lista_csv
```

```python
len(lista_csv) ## Como vemos hay 83 ficheros csv!
```

```python
## Finalmente, es interesante eliminar la parte común en 2 rutas. Puede ser útil para varias aplicaciones.
fichero_ruta_desde_base_comun = fichero.relative_to(datos) ## A la ruta "fichero" quitale toda la parte de la ruta que también esté en "datos". (de hecho lo que se hace aquí es el ".parts", crea 2 tuplas y elimina los que coinciden)
fichero_ruta_desde_base_comun ## Ahora empieza desde Download... Puede ser útil si evaluamos muchos ficheros en un directorio común y cosas (ver abajo)
```

## 2 ) **IMPORT**, **MÓDULOS** Y **SYS**

Un **módulo** solo es un fichero .py que contiene código para ser usado desde otro sitio. Contiene funciones, métodos, etc. En mi caso, el archivo **src/loader.py** seria mi módulo para el proyecto principal.
Por otro lado, una librería sería como un conjunto de módulos, o una carapeta de módulos. Por ejemplo **pandas**, es una librería, llena de módulos (ficheros .py con cosas...)

Las palabras módulo, librería, clase... son un poco ambiguas y conceptuales.

![Librería, paquete, módulo, clase y objeto](img/libreria_modulo_clase.svg)

```python
import pandas as pd ## Para importar la librería/paquete completo. Como sabes, puedes simplificarle el nombre con "as <nombre>"
from pathlib import Path ## Podemos sacar solo una clase del módulo pathlib. Un conjunto de funciones de su módulo.
from src import loader  ## Esto sería para sacar NUESTRA librería. Lo que hace es buscar en la carpeta actual pero falla, porque "src" está una carpeta por encima.
```

Cuando hacemos un **import**, Python busca primero en la carpeta actual y después en las carpetas del entorno (en este caso **.venv**)

Luego está el módulo **SYS**, que da acceso a información y control del propio intérprete. Cosas como saber qué librerías busca el interprete al leer "import" en tu código, qué entorno se está ejecutando, la versión de python,
poder modificar las rutas de búsqueda, ver el sistema operativo o terminar el programa del tirón (sys.exit())

```python
import sys

print(f"path:       {sys.path}")        # carpetas donde busca los import
print(f"executable: {sys.executable}")  # qué Python se está usando
print(f"version:    {sys.version}")     # versión de Python
print(f"platform:   {sys.platform}")    # sistema operativo
print(f"argv:       {sys.argv}")        # argumentos de lanzamiento
# sys.exit()                            # terminar el programa (NO ejecutar en el notebook)
```

```python
## Para que la librería src se encuentre, podemos cambiar las rutas donde se busca el "import" con:
sys.path.append("..") ## Esto es "añadir la carpeta '..'" que es la carpeta de arriba de esta, que es la del proyecto. Ahora también busca en esa carpeta, encuentrar src y encuentra loader.py.

## Ahora sí:
from src import loader ## Importamos el módulo
```

Finalemnte, está bien explicar algunos conceptos extra:
- Cuando una carpeta queremos que sea importable, osea que pueda importarse en otro fichero y ser usada como librería / módulo, esta tiene que tener dentro un archivo **__init__.py**. Este archivo no tiene nada, simplemente indica lo que he dicho a Python. En grandes librerías sí que puede tener código.
