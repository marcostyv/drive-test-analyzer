```python
## 3) Con ".groupby(columna)" obtenemos varios df separados por el distinto valor de esa columna. Este objeto puede acceder 
## a un valor cualquiera, por ejemplo el RSRP con: ".groupby(NetworkMode)["RSRP"]". Esto serían los RSRP de cada grupo, pero aún 
## no se puede printear porque sen grupos de columnas. 

servicios = df.groupby("NetworkMode") ## ESTO DEVUELVE UNA LISTA DE PAREJAS (tuplas)! [(tecnología_1, su_tabla), 
                                      ## (tecnología_2, su_tabla)...]. En verdad no es un alosta, es un objeto de pandas (objeto GroupBy)

servicios["RSRP"].count()
## GROUP.BY SOLO PRINTEA CUANDO HAY 1 VALOR ÚNICO QUE MOSTRAR. Por ejemplo un .mean(), un .median(), etc.
```
